## 1. Pre-Training vs Fine-Tuning Overview
- **Pre-Training:** Self-supervised next-token prediction on trillions of web tokens.
- **Fine-Tuning:** Supervised task-specific alignment and loss minimization on curated domain datasets.


## 2. Practical Scenarios Where Base Models Fail Without Fine-Tuning
1. **Healthcare & Clinical Records (ICD-10):** Clinical acronyms and precise diagnostic coding.
2. **Legal & Compliance:** Statutory precision and jurisdiction-specific contractual clauses.
3. **Quantitative Finance:** Counter-intuitive financial sentiment (e.g. contracting expenses is bullish).


## 3. Full Fine-Tuning vs PEFT
PEFT trains <1% of parameters, saves 70-80% memory, and avoids catastrophic forgetting.


## 4. Practical Implementation: Load Model, Inspect Architecture, and Freeze Base Layers


```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer
import torch

model_name = "distilbert-base-uncased"
print(f"Loading {model_name}...")
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)
tokenizer = AutoTokenizer.from_pretrained(model_name)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total Parameters     : {total_params:,}")
print(f"Trainable Parameters : {trainable_params:,} ({trainable_params/total_params*100:.2f}%)")

# Freeze Base Layers
for param in model.distilbert.parameters():
    param.requires_grad = False

total_after = sum(p.numel() for p in model.parameters())
trainable_after = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Trainable After Freezing: {trainable_after:,} ({trainable_after/total_after*100:.2f}%)")
print(f"Frozen Base Weights     : {total_after - trainable_after:,}")
```

## 5. When NOT to Fine-Tune (Decision Matrix)
- Use **Prompt Engineering** when tasks are general, sample size is low, or quick zero-shot prototyping is needed.
- Use **RAG** when data changes frequently (live news, inventory, live sports), or exact citations are required.
- Use **Fine-Tuning** when domain vocabulary, strict syntax/JSON formatting, or low latency on small models is required.
